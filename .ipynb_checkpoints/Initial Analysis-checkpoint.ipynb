### Creating column for just the years of installation

```python
#create a column for just the years of installation
heatpump_df['Year'] = pd.to_datetime(heatpump_df['Installation Start Date'], errors = 'coerce').dt.year.astype('Int64')
heatpump_df
```

### Aggregate dataset on years

```python
heatpump_df[['Customer County', 'Product Type', 'Installation Start Date']].groupby(['Customer County', 'Installation Start Date'])['Product Type'].count()

pd.set_option('display.max_rows', None)
heatpump_df.groupby(['Customer County', 'Year'])['Product Type'].count()
```

```python
#number of counties

heatpump_df['Customer County'].nunique()
```

```python
#get just the years with the highest counts of HPWHs to see if even the highest amount is significant compared to city size
counts = heatpump_df.groupby(['Customer County', 'Year'])['Product Type'].count().reset_index(name='count')
highest_per_county = (counts.sort_values(by='count', ascending = False).groupby('Customer County').first().reset_index())
highest_per_county

#sort in descending order of counts of HPWH
highest_per_county.sort_values(by = 'count', ascending = False)
```

### Obtain CA County Household Count Data

```python
#note: the following code has been written with the assistance of Google LLM Gemini
#due to storage and API access issues due to network blocks on Datahub, I used a LLM to stream data from Census Table B11001 and compile the results
#source = U.S. Census Bureau, American Community Survey 5-Year Estimates, Table B11001 (Total Households by County, California).
#numbers are from the 2022 five year release dataset

import io
import pandas as pd

# Embedded dataset of CA counties and total households (ACS estimates)
csv_data = """Customer County,Total Households
Alameda,592160
Alpine,508
Amador,15285
Butte,91799
Calaveras,20173
Colusa,7222
Contra Costa,415951
Del Norte,9851
El Dorado,75474
Fresno,318357
Glenn,10352
Humboldt,53511
Imperial,56272
Inyo,7459
Kern,283525
Kings,42371
Lake,26815
Lassen,9410
Los Angeles,3474345
Madera,46908
Marin,107248
Mariposa,7780
Mendocino,34241
Merced,77914
Modoc,3736
Mono,4985
Monterey,131379
Napa,52093
Nevada,42131
Orange,1032393
Placer,154245
Plumas,8875
Riverside,933756
Sacramento,581335
San Benito,20970
San Bernardino,734125
San Diego,1194210
San Francisco,374787
San Joaquin,231505
San Luis Obispo,113426
San Mateo,276228
Santa Barbara,146741
Santa Clara,689626
Santa Cruz,100188
Shasta,71064
Sierra,1484
Siskiyou,18032
Solano,154345
Sonoma,198595
Stanislaus,184347
Sutter,33469
Tehama,24031
Trinity,5437
Tulare,142713
Tuolumne,21568
Ventura,291415
Yolo,78567
Yuba,26960"""

# Load into a pandas DataFrame
ca_households = pd.read_csv(io.StringIO(csv_data))

ca_households
```

### Merge Household Counts by County and HPWH Installation Counts by County

```python
#note: taking some liberties with the exact years since this analysis is just to access viability on an approximate scale

#getting rid of the word "County" from the highest_per_county dataset so we can cleanly merge
highest_per_county['Customer County'] = highest_per_county['Customer County'].str.replace(r'\s+County$', '', regex = True)

merged = highest_per_county.merge(ca_households, on = 'Customer County')
merged = merged.rename(columns = {'count':'Count'})
merged
```

### Find percentage of households that installed HPWH in each county

```python
merged['HPWH Installation %'] = merged['Count'] / merged['Total Households'] * 100
merged.sort_values(by = 'HPWH Installation %', ascending = False)
```
